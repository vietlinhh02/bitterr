```python
# Tải xuống kết quả
from google.colab import files

# Danh sách các file kết quả để tải xuống
result_files = [
    '/content/ml_models/evaluation_results/evaluation_results.csv',
    '/content/ml_models/evaluation_results/confusion_matrix.png',
    '/content/ml_models/evaluation_results/class_stats.csv',
    '/content/ml_models/evaluation_results/top_accuracy.png',
    '/content/ml_models/evaluation_results/confidence_distribution.png'
]

# Tải xuống từng file
import os
for file_path in result_files:
    if os.path.exists(file_path):
        files.download(file_path)
        print(f"Đã tải xuống: {file_path}")
    else:
        print(f"File không tồn tại: {file_path}")
```
